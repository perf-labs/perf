```python
from keystone import Ks, KS_ARCH_X86, KS_MODE_64
import ctypes

df = perf.benchmark(
    symbol=(LIBC, "qsort"),
    mode=["latency"],
    event=["cycles", "instructions"],
    data={
        "regs": {
            "rsi": [1, 10, 100, 1000],
            "rdx": ctypes.sizeof(ctypes.c_int),
            "rcx": 0x4200000000,
        },
        "mem": {
            "0x4200000000:": bytes(
                Ks(KS_ARCH_X86, KS_MODE_64).asm("""
                    mov eax, dword ptr [rdi]
                    cmp eax, dword ptr [rsi]
                    setg al
                    setl dl
                    movzx eax, al
                    movzx edx, dl
                    sub eax, edx
                    ret
                """, addr=0x4200000000)[0]
            ),
        },
    },
)

summary = df.groupby("data.rsi")[["cycles", "instructions"]].median()
summary["cycles/element"] = summary["cycles"] / summary.index
summary["instructions/element"] = summary["instructions"] / summary.index
display(summary.round(2))

perf.plot(
    df,
    type=["bar"],
    event="cycles",
    groupby=["data.rsi"],
    config={"logy": True},
)
```
