```python
perf.system.tune()
```

```python
LIBC = "/lib/x86_64-linux-gnu/libc.so.6"
```

```python
display(perf.analyze(symbol=[LIBC, "qsort"]).style)
```

```python
perf.plot(
    perf.benchmark(
        symbol=[LIBC, "qsort"],
        mode=["latency"],
        event=["topdown*"],
    )
)
```

```python
perf.plot(
    perf.benchmark(
        symbol=[LIBC, "qsort"],
        mode=["latency"],
        event=["cycles"],
        data={"rdi":[1, 100, 1000, 10000]},
    ),
    type=["bar"],
    groupby=["data.rdi"],
) 
```
